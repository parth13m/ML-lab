```python
# Question p
print("Question p\n")

x = tips['total_bill'].head(5)
y = tips['tip'].head(5)

# 1. figure() & plot()
plt.figure(figsize=(5,3))
plt.plot(x,y,label="tip")
plt.title("line plot of Tip Vs Total Bill")
plt.xlabel("Total bill")
plt.ylabel("Tip")
plt.grid(True)
plt.legend()
plt.show()
# 2. scatter() & bar()
plt.figure(figsize = (5,3))
plt.scatter(x,y,color="green",label="Data Points")
plt.title("ScatterPlot")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(5, 3))
plt.bar(x, y, alpha=0.3, label='Bar Representation')
plt.title("Bar Plot")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.grid(True)
plt.legend()
plt.show()

# 3. barh() & hist()
plt.figure(figsize=(5, 3))
plt.barh(x, y, color='green', alpha=0.6)
plt.title("Horizontal Bar Plot")
plt.xlabel("Tip")
plt.ylabel("Total Bill")
plt.show()

plt.figure(figsize=(5, 3))
plt.hist(tips['tip'], bins=5, color='orange', edgecolor='black')
plt.title("Histogram of Tip")
plt.xlabel("Tip Range")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()

# 4. pie() & boxplot()
day_data = tips.groupby('day')['total_bill'].sum()
plt.figure(figsize=(5, 3))
plt.pie(day_data, labels=day_data.index, autopct='%1.1f%%')
plt.title("Total Bill Distribution by Day")
plt.show()


plt.figure(figsize=(5, 3))
plt.boxplot(tips['tip'])
plt.title("Boxplot of Tip")
plt.ylabel("Tip Values")
plt.grid(True)
plt.show()

# 5. violinplot() & imshow()
plt.figure(figsize=(5, 3))
plt.violinplot(tips['tip'])
plt.title("Violinplot of Tip")
plt.ylabel("Tip Values")
plt.grid(True)
plt.show()

plt.figure(figsize=(5, 3))
plt.imshow(tips[['total_bill', 'tip']].head(30), cmap='viridis', aspect='auto')
plt.title("Imshow Matrix Display")
plt.colorbar()
plt.show()

# 6. subplot()
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 3))
ax1.plot(x, y)
ax1.set_title("Subplot 1: Line")
ax2.scatter(x, y, color='red')
ax2.set_title("Subplot 2: Scatter")
plt.tight_layout()
plt.show()
```

```python
# Question r
print("Question r")
import numpy as np
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import (
    GridSearchCV,
    KFold,
    RandomizedSearchCV,
    StratifiedKFold,
    cross_val_score,
)
from sklearn.svm import SVC

X, y = load_iris(return_X_y=True)
# 1. KFold
kf = KFold(n_splits=5, shuffle=True, random_state=42)
for train_idx, test_idx in kf.split(X):
    pass

# 2. StratifiedKFold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
for train_idx, test_idx in skf.split(X, y):
    pass

# 3. cross_val_score
cv_scores = cross_val_score(
    LogisticRegression(max_iter=200), X, y, cv=5, scoring="accuracy"
)
print("Cross-Validation Mean Accuracy:", cv_scores.mean())

# 4. GridSearchCV
grid_param = {"C": [0.1, 1, 10], "kernel": ["linear", "rbf"]}
grid_search = GridSearchCV(
    estimator=SVC(), param_grid=grid_param, cv=5, scoring="accuracy"
)
grid_search.fit(X, y)
print("GridSearchCV Best Parameters:", grid_search.best_params_)

# 5. RandomizedSearchCV
random_param = {"n_estimators": [10, 50, 100], "max_depth": [None, 5, 10]}
random_search = RandomizedSearchCV(
    estimator=RandomForestClassifier(),
    param_distributions=random_param,
    n_iter=5,
    cv=5,
    random_state=42,
)
random_search.fit(X, y)
print("RandomizedSearchCV Best Parameters:", random_search.best_params_)
```
